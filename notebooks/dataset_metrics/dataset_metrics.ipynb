```python
from dataset_readers import *
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from math import isnan
import scipy

from aif360.metrics import ClassificationMetric
from models import SimpleMLP
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import matthews_corrcoef as mcc

from xicorrelation import xicorr
from scipy.stats import spearmanr, pearsonr, kendalltau
import matplotlib as mpl
mpl.rcParams['figure.dpi'] = 300
sns.set_style('whitegrid')
```

```python
datasets = [
    adult_dataset_reader,
    bank_dataset_reader,
    compas_dataset_reader,
    german_dataset_reader
]
```

```python
def get_sorted_corr(dataset, correlation_function, target_feature_index):
    correlations = []
    target_feature = dataset[:, target_feature_index]

    for i in range(dataset.shape[1]):
        corr, _ = correlation_function(target_feature, dataset[:, i])
        if not isnan(corr):
            correlations.append(abs(corr))
    
    sorted_correlations = sorted(correlations, reverse=True)
    
    return sorted_correlations

```

```python
dataset_info = []
datasets_names = ['Adult', 'Bank', 'Compas', 'German']
correlation_functions = [pearsonr, spearmanr, kendalltau, xicorr]
correlation_names = ['Pearson $r$', 'Spearman $\\rho$', 'Kendall $\\tau$', 'Chatterjee $\\xi$']
for correlation_function, correlation_name in zip(correlation_functions, correlation_names):
    for dataset_reader, dataset_name in zip(datasets, datasets_names):
        dataset_expanded_train, dataset_train, dataset_val, dataset_test, unprivileged_groups, privileged_groups, sens_attr = dataset_reader(shuffle=False)
    
        sensitive_index = dataset_expanded_train.protected_attribute_names.index(sens_attr)
        print(dataset_name, correlation_name, sens_attr, sensitive_index)
        sorted_corrs = get_sorted_corr(
            np.vstack([dataset_expanded_train.features, dataset_test.features]) 
                          , correlation_function, sensitive_index)
        dataset_info.append({
            'correlation_function': correlation_name,
            'sorted_corrs': sorted_corrs,
            'dataset_name': dataset_name  # Nome do dataset, ajuste conforme necessário
        })
        
```

```python
def set_x_limits(ax, xlim, color=None, **kwargs):
    if xlim == '1-10':
        ax.set_xlim(1, 10)
    elif xlim == '1-400':
        ax.set_xlim(1, 400)
```

```python
plot_data = []
for info in dataset_info:
    for index, corr in enumerate(info['sorted_corrs']):
        plot_data.append({
            'correlation_function': info['correlation_function'],
            'feature_index': index,
            'correlation_value': corr,
            'dataset_name': info['dataset_name'],
            'xlim': '1-10'
        })
        plot_data.append({
            'correlation_function': info['correlation_function'],
            'feature_index': index,
            'correlation_value': corr,
            'dataset_name': info['dataset_name'],
            'xlim': '1-400'
        })

plot_df = pd.DataFrame(plot_data)
```

```python
correlation_names = ['Pearson $r$', 'Spearman $\\rho$', 'Kendall $\\tau$', 'Chatterjee $\\xi$']
datasets_names = ['Adult', 'Bank', 'Compas', 'German']
viridis_colors = sns.color_palette('viridis', n_colors=len(datasets_names))

# Plot 1: four correlation types, range 1-10, in a single row
fig1, axes1 = plt.subplots(1, 4, figsize=(11, 3), sharey=True)

handles, labels = None, None

for i, correlation_name in enumerate(correlation_names):
    ax = axes1[i]
    subset = plot_df[(plot_df['correlation_function'] == correlation_name) & (plot_df['xlim'] == '1-10')]
    sns.lineplot(
        data=subset,
        x='feature_index',
        y='correlation_value',
        hue='dataset_name',
        ax=ax,
        palette=viridis_colors,
        linewidth=1.5
    )
    if handles is None and labels is None:
        handles, labels = ax.get_legend_handles_labels()
    ax.set_xlim(1, 10)
    ax.set_title(correlation_name)
    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.legend_.remove()

axes1[0].set_ylabel('correlation')
fig1.text(0.5, 0.05, r'$n$-th most correlated feature to sensitive attribute', ha='center')

fig1.legend(handles, labels, loc='lower center', bbox_to_anchor=(0.5, -0.12),
            ncol=len(labels), fontsize='small', frameon=False)

plt.tight_layout(rect=[0, 0.08, 1, 1])
plt.savefig('plots/dataset_correlation_top10.png', bbox_inches='tight')
plt.savefig('plots/dataset_correlation_top10.pdf', bbox_inches='tight')
plt.show()

# Plot 2: only Chatterjee, range 1-400
fig2, ax2 = plt.subplots(1, 1, figsize=(6, 4))

subset_chatterjee = plot_df[
    (plot_df['correlation_function'] == 'Chatterjee $\\xi$') &
    (plot_df['xlim'] == '1-400')
]
sns.lineplot(
    data=subset_chatterjee,
    x='feature_index',
    y='correlation_value',
    hue='dataset_name',
    ax=ax2,
    palette=viridis_colors,
    linewidth=1.5
)
ax2.set_xlim(1, 400)
ax2.set_xlabel( r'$n$-th most correlated feature to sensitive attribute')
ax2.set_ylabel('Chatterjee $\\xi$')

plt.tight_layout()
plt.savefig('plots/dataset_correlation_chatterjee400.png', bbox_inches='tight')
plt.savefig('plots/dataset_correlation_chatterjee400.pdf', bbox_inches='tight')
plt.show()
```
