```python
def read_csv_files_from_folder(folder_path):
    # Initialize an empty list to store DataFrames
    dfs = []
    # Get a list of files in the folder
    file_list = os.listdir(folder_path)
    # Iterate through the files in the folder
    for file_name in file_list:
        # Check if the file has a .csv extension
        if file_name.endswith(".csv"):
            file_path = os.path.join(folder_path, file_name)

            # Read the CSV file into a DataFrame
            df = pd.read_csv(file_path)

            # Append the DataFrame to the list
            dfs.append(df)

    # Concatenate all DataFrames into a single DataFrame
    combined_df = pd.concat(dfs, ignore_index=True)
    return combined_df
```

```python
mlp_baseline_results = pd.read_csv('simple_mlp_results.csv')
mlp_baseline_results.replace({'simple_mlp_initializer': 'MLP'}, inplace=True)

mlp_xi_reg_results = pd.read_csv('mlp_xi_reg_results.csv')
mlp_xi_reg_results.replace({'mlp_xi_reg_initializer': r'MLP+SDR$_{\xi}$'}, inplace=True)

mlp_results = pd.concat([mlp_baseline_results, mlp_xi_reg_results])

ftl_baseline_results = pd.read_csv('ftl_mlp_results.csv')
ftl_baseline_results.replace({'ftl_mlp_initializer': 'FTL'}, inplace=True)

ftl_xi_reg_results = pd.read_csv('ftl_mlp_xi_reg_results.csv')
ftl_xi_reg_results.replace({'ftl_mlp_xi_reg_initializer': r'FTL+SDR$_{\xi}$'}, inplace=True)

ftl_results = pd.concat([ftl_baseline_results, ftl_xi_reg_results])

# HIFI results are stored as raw per-run CSVs produced by ablation.py's hifi_initializer
_hifi_cols = ['dataset', 'method', 'fitness_rule', 'fitness', 'ACC', 'MCC',
              'avg_odds_diff', 'stat_par_diff', 'eq_opp_diff', 'solution']
hifi_raw_results = read_csv_files_from_folder('raw/')
hifi_results = hifi_raw_results[hifi_raw_results['method'] == 'hifi_initializer'][_hifi_cols].copy()
hifi_results.replace({'hifi_initializer': 'HIFI'}, inplace=True)

results = pd.concat([mlp_results, ftl_results, hifi_results])
```

## Possible names
- Sensitive Correlation Regularization/Penalty
- Redlining Regularization/Penalty
- Senstive-$L_2$ Regularization

Let
- $X \in \mathbb{R}^{n\times d}$ represent a dataset with $n$ instances and $d$ features
- $X_i \in \mathbb{R}^n, \; 1 \leq i \leq d$ be the $i$-th feature of the dataset
- $A = X_i \in \mathbb{R}^n$ the sensitive feature (protected) for some $i$
- $W^{(1)} \in \mathbb{R}^{d \times h}$ the weight matrix for the first hidden layer, with $h$ being the number of neurons in this layer.
- $\lambda \in \mathbb{R}^{d}$ the vector of regularization strengths for each feature
- $\alpha$ a scalar that controls the overall strength of the regularization

### Regularization Term
The regularization term $R(\mathbf{W}^{(1)})$ applied to the weight matrix $\mathbf{W}^{(1)}$ of the first hidden layer is defined as:
$$R(\mathbf{W}^{(1)}) = \sum_{i=1}^d \lambda_i \sum_{j=1}^h (W^{(1)}_{ij})^2$$
Here, $\lambda_i = |correlation(A,X_i)|$ is the regularization strength for the $i$-th input feature using an arbitrary correlation coefficient (eg. Pearson r, Spearman r, Kendall tau, Chatterjee xi) within sensitive feature A, and $W^{(1)}_{ij}$ is the weight connecting the $i$-th input feature to the $j$-th neuron in the first hidden layer.

The total loss function $L$ for the MLP, incorporating sensitive-$L_2$ regularization, is:
$$L = L_0 + \alpha \; R(\mathbf{W}^{(1)})$$
where $L_0$ is the primary loss function of the network




```python
# 4 datasets x 6 fitness rules
performance_metrics = ['mcc', 'acc']
fairness_metrics = ['parity', 'opportunity', 'odds']

# Iterate through datasets (rows) and fitness rules (columns)
for i, performance_metric in enumerate(performance_metrics):

    fig, axes = plt.subplots(nrows=4, ncols=3, figsize=(8, 5), sharex=False)
    fig.subplots_adjust(hspace=0.0, wspace=0.3)

    for j, dataset in enumerate(datasets):
        for k, fairness_metric in enumerate(fairness_metrics):
            ax = axes[j, k]

            fitness_rule = f"{performance_metrics[i]}_{fairness_metrics[k]}"
            # Filter data for current dataset and fitness rule
            subset = results[(results.dataset == dataset) & (results.fitness_rule == fitness_rule)]
            subset['method'] = pd.Categorical(subset['method'], categories=methods)
            subset = subset.sort_values(by='method')

            # Create boxplot
            boxplot = sns.boxplot(data=subset, hue='method', x='fitness', y='method', ax=ax, palette=method_color_map)

            # Set titles and labels
            if j == 0:  # First row - add fitness rule titles
                ax.set_title(fitness_rules_target_metrics[fitness_rule][1], fontsize=10, pad=10)
            else:
                ax.set_title('')

            ax.yaxis.set_label_position("right")
            #ax.yaxis.label.set_rotation(270)
            if k == len(fairness_metrics) - 1:  # Last column - add dataset labels
                ax.set_ylabel(dataset, fontsize=10)
            else:
                ax.set_ylabel('')

            # Remove x-axis labels except for bottom row
            if j < len(datasets) - 1:
                ax.set_xlabel('')
            else:
                ax.set_xlabel('Fitness', fontsize=10)

            # Remove y-axis tick labels except for first column
            if k > 0:
                ax.set_yticklabels([])

    # Add a single legend for the entire figure
    # handles, labels = axes[0, 0].get_legend_handles_labels()
    # fig.legend(handles, labels, loc='upper center', bbox_to_anchor=(0.5, 0.95), ncol=4, fontsize=12)

    plt.tight_layout()
    plt.subplots_adjust(top=0.92)  # Make room for the legend
    plt.savefig(f"plots/boxplot_{performance_metric}.pdf", bbox_inches='tight')
    plt.savefig(f"plots/boxplot_{performance_metric}.png", bbox_inches='tight')
    plt.show()
```

```python
# Marker/color assignment kept consistent with notebooks/ablation_results/plot_sensibility.ipynb
methods_all = [r'MLP', r'MLP+SDR$_{\xi}$', r'FTL', r'FTL+SDR$_{\xi}$', r'HIFI']
method_markers = ['o', 's', '^', 'D', 'v', 'P', 'X']
method_marker_map = {m: method_markers[i % len(method_markers)] for i, m in enumerate(methods_all)}

cmap = plt.cm.RdBu_r
method_palette = [cmap(0.15), cmap(0.25), cmap(0.75), cmap(0.85), 'tab:green']
method_color_map = {m: method_palette[i % len(method_palette)] for i, m in enumerate(methods_all)}

def _pad_limits(values, frac=0.05):
    values = list(values)
    if not values:
        return None
    lo, hi = min(values), max(values)
    if lo == hi:
        return lo - 0.05, hi + 0.05
    margin = (hi - lo) * frac
    return lo - margin, hi + margin
```

```python
# Scatter grid: rows = fairness metrics, columns = datasets, split into
# two 4-column blocks (left: MCC, right: Accuracy), following the same
# layout used for the lambda sensibility plot in plot_sensibility.ipynb

os.makedirs('plots', exist_ok=True)

TOP_N = 5  # keep only the top-N results by fitness per method/dataset/objective

blocks = [
    ('Mathew Correlation', ['mcc_parity', 'mcc_opportunity', 'mcc_odds']),
    ('Accuracy', ['acc_parity', 'acc_opportunity', 'acc_odds']),
]

n_rows = len(blocks[0][1])
n_lambda_like = len(datasets)
n_cols = n_lambda_like * 2

def top_n_by_fitness(df, n=TOP_N):
    return df.sort_values(by='fitness', ascending=False).head(n)

# Shared x-limits per block, shared y-limits per row (fairness metric),
# computed only over the top-N filtered points that will actually be plotted
block_xlims = []
for perf_label, rules in blocks:
    perf_values = []
    for rule in rules:
        performance_metric, _ = fitness_rules_target_metrics[rule]
        rule_subset = results[results.fitness_rule == rule]
        for dataset in datasets:
            for method in methods:
                grp = rule_subset[(rule_subset.dataset == dataset) & (rule_subset.method == method)]
                perf_values += list(top_n_by_fitness(grp)[performance_metric])
    block_xlims.append(_pad_limits(perf_values))

row_ylims = []
for row in range(n_rows):
    fairness_values = []
    for perf_label, rules in blocks:
        rule = rules[row]
        _, fairness_metric = fitness_rules_target_metrics[rule]
        rule_subset = results[results.fitness_rule == rule]
        for dataset in datasets:
            for method in methods:
                grp = rule_subset[(rule_subset.dataset == dataset) & (rule_subset.method == method)]
                fairness_values += list(top_n_by_fitness(grp)[fairness_metric])
    row_ylims.append(_pad_limits(fairness_values))

fig, axes = plt.subplots(nrows=n_rows, ncols=n_cols, figsize=(2 * n_cols, 2 * n_rows))

for b, (perf_label, rules) in enumerate(blocks):
    xlim = block_xlims[b]
    for row, rule in enumerate(rules):
        performance_metric, fairness_metric = fitness_rules_target_metrics[rule]
        ylim = row_ylims[row]
        for col, dataset in enumerate(datasets):
            global_col = b * n_lambda_like + col
            ax = axes[row, global_col]
            subset = results[(results.dataset == dataset) & (results.fitness_rule == rule)]

            for method in methods:
                method_subset = subset[subset['method'] == method]
                method_subset = top_n_by_fitness(method_subset)
                ax.scatter(method_subset[performance_metric], method_subset[fairness_metric],
                           color=method_color_map[method],
                           marker=method_marker_map[method])

            if xlim is not None:
                ax.set_xlim(xlim)
            if ylim is not None:
                ax.set_ylim(ylim)

            if row == 0:
                ax.set_title(dataset)

            if global_col == 0:
                ax.set_ylabel(fairness_metric)
            else:
                ax.set_yticklabels([])

            if row == n_rows - 1:
                ax.set_xlabel(perf_label)
            else:
                ax.set_xticklabels([])

plt.tight_layout(rect=(0, 0.08, 1, 1))

legend_handles = [Line2D([0], [0], marker=method_marker_map[m], color=method_color_map[m],
                          linestyle='', markersize=8, label=m) for m in methods]
fig.legend(handles=legend_handles, loc='lower center', ncol=len(methods),
           bbox_to_anchor=(0.5, 0.0), title='Model')

plt.savefig('plots/scatter_sensibility_combined.pdf', bbox_inches='tight')
plt.savefig('plots/scatter_sensibility_combined.png', bbox_inches='tight')
plt.show()
```
